```python
def preprocess(text: str): 
    """Lowercase, strip punctuation, tokenize, and lemmatize a string."""
    text = text.lower().translate(str.maketrans("", "", string.punctuation))
    tokens = nltk.word_tokenize(text)
    lemmas = [lemmatizer.lemmatize(tok) for tok in tokens]
    return lemmas

print(preprocess("Hello! Who are you?"))
```

```python
processed_patterns = {
    tag: [preprocess(p) for p in data["patterns"]]
    for tag, data in intents.items()
}


def score_intent(user_tokens, pattern_tokens):
    """Overlap score between user tokens and a single pattern's tokens."""
    if not pattern_tokens:
        return 0.0
    overlap = len(set(user_tokens) & set(pattern_tokens))
    return overlap / len(set(pattern_tokens))


def match_intent(user_text, threshold=0.8):
    user_tokens = preprocess(user_text)
    best_tag, best_score = "fallback", 0.0

    for tag, pattern_list in processed_patterns.items():
        if tag == "fallback":
            continue
        for pattern_tokens in pattern_list:
            score = score_intent(user_tokens, pattern_tokens)
            if score > best_score:
                best_score, best_tag = score, tag

    if best_score < threshold:
        return "fallback"
    return best_tag


def get_response(user_text):
    tag = match_intent(user_text)
    return random.choice(intents[tag]["responses"])


# Quick test
for test in ["hello there", "thanks", "what is your name", "jdfhieufjduh"]:
    print(f"{test!r:30s} -> {get_response(test)}")
```

```python
def run_text_chatbot():
    print("Text chatbot started. Type 'bye' to stop.\n")
    while True:
        user_text = input("You: ")
        tag = match_intent(user_text)
        response = get_response(user_text)
        print(f"Bot: {response}")
        if tag == "goodbye":
            break

run_text_chatbot()
```
