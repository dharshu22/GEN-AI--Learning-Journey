```python
import spacy
from spacy.tokens import Doc
nlp = spacy.load("en_core_web_sm")

positive = ["good", "happy", "excellent", "love", "great"]
negative = ["bad", "sad", "hate", "worst", "poor"]

def sentiment(doc):
    score = 0
    for token in doc:
        if token.text.lower() in positive:
            score += 1
        elif token.text.lower() in negative:
            score -= 1
    return score

Doc.set_extension("sentiment", getter=sentiment, force=True)

doc = nlp("I love this movie")
doc1 = nlp("I hate this")
doc2 = nlp(" Why are you so sad?")
print(doc._.sentiment)
print(doc1._.sentiment)
print(doc2._.sentiment)

```

```python

positive = ["good", "happy", "love", "great"]
negative = ["bad", "sad", "hate", "terrible"]

def emotion(doc):
    score = 0

    for token in doc:
        if token.text.lower() in positive:
            score += 1
        elif token.text.lower() in negative:
            score -= 1

    if score > 0:
        sentiment = "Positive"
    elif score < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    return {
        "Polarity Score": score,
        "Sentiment": sentiment
    }

Doc.set_extension("emotion", getter=emotion, force=True)

doc1 = nlp("I hate this game")
doc2 = nlp("I love this game")
doc3 = nlp("It's terrible")
print(doc1._.emotion)
print(doc2._.emotion)
print(doc3._.emotion)
```
