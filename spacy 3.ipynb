```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc2=nlp("The Indian Kho-Kho team won the championship in Chennai on 15 August 2025 under coach Rajesh Kumar.")
for i in doc2.ents:
    print(i.text,i.label_)
```

```python
from collections import Counter 
nlp=spacy.load('en_core_web_sm')
text ="The Indian Kho-Kho team won the championship in Chennai on 15 August 2025 under coach Rajesh Kumar." 
doc=nlp(text)
wordfrequency=Counter(doc)
print(wordfrequency)
```

```python
common_word=wordfrequency.most_common(5)
print(common_word)
```

```python
import spacy
from collections import Counter
nlp=spacy.load('en_core_web_sm')
text = "The Indian Kho-Kho team won the championship in Chennai on 15 August 2025 under coach Rajesh Kumar."
doc=nlp(text)
words=[t.text for t in doc if t.is_stop!=True and t.is_punct!=True] 
wordfrequency=Counter(words)

print(wordfrequency)
```
